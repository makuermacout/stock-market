## 1. Key concepts

A strong idea links four things: **opportunity → thesis → data → test**.

- **Weak:** "Gaming is growing, so buy gaming stocks."
- **Strong:** "Player counts rise ahead of earnings; analysts update slowly, so revenue surprises are predictable. Test post-earnings returns."

**Edge** = a reason the price is slow or wrong: limited coverage, information processing (local language, niche data), behavioral overreaction, structural relationships. Growth alone is *not* an edge.

| Idea | Core mechanism | What kills it |
|---|---|---|
| Pairs trading | Spread of related stocks mean-reverts | Structural break, drifting hedge ratio |
| Mean reversion | Overreaction reverses | Move was a lasting business change |
| Local markets | Slow information absorption | Thin liquidity, wide spreads, can't short |
| Sector demand signals | Alternative data predicts revenue surprises | Already priced in, few observations |

**Validation checklist:** benchmark, out-of-sample test, rejection criteria set in advance, frictions (costs, slippage, liquidity, publication delay, look-ahead bias).

## 3. Self-test answers

**Q1. Why is "the industry is growing" not an investment thesis?**
Growth is public, so it is already priced in. A thesis must state what information predicts which outcome, over what horizon, and why the market has not priced it. A growing industry can still give poor returns if expectations were higher, competition erodes margins, or the stock is expensive. Ask what your signal reveals *beyond* the growth story.

**Q2. Temporary overreaction vs. lasting change: how to tell them apart?**
- *Overreaction:* price move larger than the news justifies (sentiment, forced selling); tends to reverse.
- *Lasting change:* real shift (lost customer, regulation, margin change); new price is justified.

Tests: (a) fundamentals after the event (revenue, margins, guidance intact?); (b) volume and news (spike with no new information or a one-off cause); (c) peer behavior (company-specific vs. sector-wide); (d) historical analogues for similar events, tested out-of-sample.

**Q3. Three reasons a backtest profit can disappear live.**
1. Transaction costs and slippage (spreads, commissions, price impact), especially in illiquid stocks.
2. Overfitting: parameters tuned to past data captured noise.
3. Look-ahead bias or data issues: late-published data, restated figures, survivorship bias.

Also: regime change, crowding, inability to short or fill at assumed prices.

**Q4. Why does cointegration testing on the data used to pick the pair risk overfitting?**
Scanning many pairs and keeping those that look cointegrated is a multiple-testing problem: some pass by chance. The relationship may be a statistical accident that breaks out-of-sample. Fix: pick pairs on economic logic, estimate on a training window, evaluate on a separate test window the selection never saw.

**Q5. Publication delay and look-ahead bias.**
Publication delay is the gap between the period a data point describes and the date it becomes public. Look-ahead bias arises when a backtest uses the figure from the period it describes rather than its release date, so it trades on information nobody had. Fix: timestamp each point with its release date, use it only from then, and use as-first-reported values where data are revised.

**Exam pattern:** one-line concept → mechanism → concrete failure mode.

## 4. Demo: pairs-trading z-score (synthetic data)

This uses simulated prices (no downloads), so it runs offline. It illustrates the mechanics only: hedge ratio, spread, z-score, entry/exit signals, and why you test out-of-sample. It is **not** evidence of a profitable strategy.

```python
import numpy as np
import pandas as pd

rng = np.random.default_rng(42)
n = 1000

# Two related log-price series sharing a common trend, plus a mean-reverting gap
common = np.cumsum(rng.normal(0.0003, 0.01, n))
gap = np.zeros(n)
for t in range(1, n):
    gap[t] = 0.95 * gap[t-1] + rng.normal(0, 0.01)   # AR(1): mean-reverting

a = 4.0 + common + 0.5 * gap + rng.normal(0, 0.002, n)
b = 4.0 + common - 0.5 * gap + rng.normal(0, 0.002, n)
df = pd.DataFrame({"A": a, "B": b})

# Train / test split: estimate on train only, evaluate on test
split = 700
train, test = df.iloc[:split], df.iloc[split:]

# Hedge ratio from training window only
beta = np.polyfit(train["B"], train["A"], 1)[0]
spread_train = train["A"] - beta * train["B"]
mu, sd = spread_train.mean(), spread_train.std()

# Apply frozen parameters to test window
spread_test = test["A"] - beta * test["B"]
z = (spread_test - mu) / sd

# Signals: short spread when z>2, long when z<-2, exit near 0
pos = pd.Series(0.0, index=z.index)
state = 0
for i, val in enumerate(z):
    if state == 0:
        if val > 2: state = -1
        elif val < -2: state = 1
    elif abs(val) < 0.2 or abs(val) > 3.5:   # exit or stop
        state = 0
    pos.iloc[i] = state

# Simple P&L on spread changes (ignores costs: add them in a real test!)
pnl = pos.shift(1) * spread_test.diff()
print(f"hedge ratio (train): {beta:.3f}")
print(f"trades entered: {(pos.diff().abs() > 0).sum()}")
print(f"cumulative spread P&L (no costs): {pnl.sum():.4f}")
print("Next steps: add costs, test cointegration, vary thresholds, compare to random pairs.")
```

**Takeaways from the demo**
- Parameters (beta, mean, std) are estimated on the training window and frozen for the test window, which avoids look-ahead bias.
- Real prices have structural breaks and costs; this synthetic spread is mean-reverting by construction, so results here are optimistic.
